```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/kaggle/input/datasets/ayushish12/positionsalaries-data/position-salaries data.csv')

df.head()
```

```python
X = df['Level'].values.reshape(-1, 1)
y = df['Salary'].values.reshape(-1, 1)
```

```python
X_mean = np.mean(X)
X_std = np.std(X)

X_scaled = (X - X_mean) / X_std

X_poly = np.hstack((
    np.ones((len(X), 1)),
    X_scaled,
    X_scaled**2,
    X_scaled**3,
    X_scaled**4
))

y_scaled = y / 100000
```

```python
n = len(X)
m_features = X_poly.shape[1]

weights = np.zeros((m_features, 1))

learning_rate = 0.01
epochs = 10000

beta = 0.9

velocity = np.zeros_like(weights)

loss_history = []
```

```python
for i in range(epochs):

    # Prediction
    y_pred = X_poly @ weights

    # Error
    error = y_pred - y_scaled

    # Gradient
    dw = (2 / n) * (X_poly.T @ error)

    # Momentum
    velocity = beta * velocity + (1 - beta) * dw

    # Update weights
    weights -= learning_rate * velocity

    # Loss
    loss = np.mean(error ** 2)
    loss_history.append(loss)

print("Final Loss:", loss_history[-1])
print("Weights:")
print(weights)
```

```python
plt.figure(figsize=(8,5))

plt.plot(loss_history)

plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.title('GD with Momentum - Loss')

plt.grid()
plt.show()
```

```python
X_curve = np.linspace(X.min(), X.max(), 200).reshape(-1, 1)

X_curve_scaled = (X_curve - X_mean) / X_std

X_curve_poly = np.hstack((
    np.ones((len(X_curve), 1)),
    X_curve_scaled,
    X_curve_scaled**2,
    X_curve_scaled**3,
    X_curve_scaled**4
))

y_curve = (X_curve_poly @ weights) * 100000

plt.figure(figsize=(8,5))

plt.scatter(X, y, color='red', label='Actual Data')

plt.plot(X_curve, y_curve, color='blue', label='GD with Momentum')

plt.xlabel('Position Level')
plt.ylabel('Salary')
plt.title('Polynomial Regression using GD with Momentum')

plt.legend()
plt.show()
```
